```python
import pandas as pd
from sklearn.impute import SimpleImputer

sample = pd.DataFrame({
    "Age": [22, None, 35, 28, None]
})

imputer = SimpleImputer(strategy="median")

sample["Age"] = imputer.fit_transform(sample[["Age"]]).ravel()

print(sample)
```

```python
import pandas as pd
from sklearn.impute import KNNImputer

sample = pd.DataFrame({
    "Age": [22, 25, None, 35, 40],
    "Fare": [10, 15, 12, 50, 60]
})

imputer = KNNImputer(n_neighbors=2)

result = imputer.fit_transform(sample)

result = pd.DataFrame(result, columns=sample.columns)

print(result)
```

```python
import pandas as pd

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

sample = pd.DataFrame({
    "Age": [22, 25, None, 35, 40],
    "Fare": [10, 15, 12, 50, 60],
    "Pclass": [3, 3, 3, 1, 1]
})

imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

result = imputer.fit_transform(sample)

result = pd.DataFrame(
    result,
    columns=sample.columns
)

print(result)
```

```python
import pandas as pd

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

sample = pd.DataFrame({
    "Age": [22, 25, None, 35, 40],
    "Fare": [10, 15, 12, 50, 60],
    "Pclass": [3, 3, 3, 1, 1]
})

imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

result = imputer.fit_transform(sample)

result = pd.DataFrame(
    result,
    columns=sample.columns
)

print(result)
```
